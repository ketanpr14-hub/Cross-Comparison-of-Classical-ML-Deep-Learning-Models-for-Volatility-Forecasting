```python
import pandas as pd, numpy as np
import matplotlib.pyplot as plt

# a notebook in notebooks/ runs from that folder, hence "../data"
path = "../data/realized_variance_stocks_20261030/realized_variance_stocks_20261030.csv"
stocks = pd.read_csv(path, parse_dates=["date"])
stocks.head()
```

```python
aapl = stocks[stocks['symbol'] == 'AAPL'].sort_values('date').reset_index(drop=True)
aapl
```

```python
aapl["y"] = np.log(aapl["rv5"])  # today's log volatility
```

```python
aapl["x_day"] = aapl["y"]  # input 1: today
aapl["x_week"] = aapl["y"].rolling(5).mean() # input 2: average of the last 5 days
aapl["x_month"] = aapl["y"].rolling(22).mean()  # input 3: average of the last 22 days
```

```python
aapl
```

```python
aapl["target"] = aapl["y"].shift(-1)             # what we predict: tomorrow's value
```

```python
aapl
```

```python
data = aapl.dropna(subset=["x_month", "target"])
data[["date", "y", "x_day", "x_week", "x_month", "target"]].head()
```

```python
data["target"].iloc[0] == data["y"].iloc[1]
```

## Train test split and training a base model

```python
train = data[data['date'] < "2019-01-01"]
test = data[data['date'] >= "2019-01-01"]
print(len(train), len(test))
```

```python
# baseline 1: "tomorrow = today"
mse_naive = ((test['target']- test['x_day'])**2).mean()

# baseline 2: always predict the average of the training period
mse_mean = ((test['target'] - train['target'].mean())**2).mean()

print(mse_naive, mse_mean)
print('tomorrow=today has lower mse and is thefroe better model')
```

```python
#what I propose as a baseline
lst = []
for c in ["x_day", "x_week", "x_month"]:
    print(c, ((test["target"] - test[c]) ** 2).mean())
    lst.append(((test["target"] - test[c]) ** 2).mean())

lst = np.array(lst)
print(lst.mean())
```

```python
pred_avg = test[["x_day", "x_week", "x_month"]].mean(axis=1)
print(((test["target"] - pred_avg) ** 2).mean())
```

## Baseline HAR simple model with sklearn

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

features = ["x_day", "x_week", "x_month"]
har = LinearRegression()
har.fit(train[features], train["target"])

pred_har = har.predict(test[features])

mse_har = mean_squared_error(test['target'], pred_har)
print(har.coef_, har.intercept_, mse_har)
```

### The main gain came from combining horizons. Going from 0.411 to 0.321 is far bigger than going from 0.321 to 0.317, which is what learning the weights added. Treat that small gap with caution. It’s one stock and one split, so it’s too small to call HAR clearly better than the plain average. We’ll test that properly later.
The weights are 0.47 on the day, 0.23 on the week and 0.20 on the month. They overlap because the week and month averages both include today, so they aren’t three independent votes.
The weights add up to about 0.90, below 1, and the intercept is −0.94. Together they pull every forecast back toward a long-run level of about −9.2 (roughly 1% daily volatility). That’s mean reversion. After a calm or wild day, volatility tends to drift back toward normal.
Scale: HAR removes about 64% of the error of the know-nothing baseline.

```python
import torch
from torch import nn

# 1. pandas -> PyTorch tensors (float32; the np.array copy avoids a "not writable" warning)
X_train = torch.tensor(np.array(train[features], dtype=np.float32))
y_train = torch.tensor(np.array(train["target"], dtype=np.float32)).reshape(-1, 1)
X_test  = torch.tensor(np.array(test[features],  dtype=np.float32))
y_test  = torch.tensor(np.array(test["target"],  dtype=np.float32)).reshape(-1, 1)

# 2. the model: one linear layer = intercept + w1*x_day + w2*x_week + w3*x_month
model = nn.Linear(3, 1)

# 3. how to measure the error, and how to change the weights
loss_fn = nn.MSELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.05)

# 4. the training loop
for epoch in range(3000):
    pred = model(X_train)           # forward pass: make predictions
    loss = loss_fn(pred, y_train)   # how wrong are we?
    optimizer.zero_grad()           # clear the previous step's gradients
    loss.backward()                 # work out how each weight should change
    optimizer.step()                # change the weights a little

# 5. score on the test set (no_grad = we're not training now)
with torch.no_grad():
    test_loss = loss_fn(model(X_test), y_test)
print(model.weight, model.bias, test_loss.item())
```

# First MLP model

```python
losses = []
for seed in [0, 1, 2]:
    torch.manual_seed(seed)
    # 2. the model:
    mlp = nn.Sequential(
            nn.Linear(3, 8),
            nn.ReLU(),
            nn.Linear(8, 1)
            )
    # 3. how to measure the error, and how to change the weights
    loss_fn = nn.MSELoss() #when will we use QLE or wtv
    optimizer = torch.optim.Adam(mlp.parameters(), lr=0.05) #RMSprop is also frequently used could try later

    for epoch in range(3000):
        pred = mlp(X_train)           # forward pass: make predictions
        loss = loss_fn(pred, y_train)   # how wrong are we?
        optimizer.zero_grad()           # clear the previous step's gradients
        loss.backward()                 # work out how each weight should change
        optimizer.step()                # change the weights a little

# 5. score on the test set (no_grad = we're not training now)
    with torch.no_grad():
        test_loss = loss_fn(mlp(X_test), y_test)

    losses.append(test_loss.item())
print(losses)
```

### The MLP ties HAR. The three seeds differ by about 0.0002, so the result is stable and not luck. That’s not a failure. Here’s why:

### The 8 hidden units found essentially a linear rule. With these three inputs, the relationship between today’s volatility summaries and tomorrow’s is close to a straight line in logs, so the extra flexibility has nothing to capture.

### This matches the published evidence. Brini’s best foundation model beat Log-HAR by under 2% on this data. A tie with a classical model is a normal outcome in volatility forecasting, and it’s worth reporting.
